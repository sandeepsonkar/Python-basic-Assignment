Q1. Print numbers from 1 to 20 using while loop.

- Initialize a variable

- Increment inside loop

Q4. Find sum of numbers from 1 to n.

Approach:
- Take input n
- Add numbers in loop

```python
#write answer here
i=1
sum = 0
num = int(input("Enter the number: "))
while(i<=num):
    sum += i
    i+=1
print("The sum is:",sum)
```

Q5. Reverse a number using while loop.

Approach:
- Extract last digit
- Build reversed number

```python
#write answer here
num = int(input("Enter the number to be reversed: "))
while(num!=0):
    digit = num % 10
    num = num // 10
    print(digit, end="")
```

Q6. Count digits in a number.

Approach:
- Divide number by 10 repeatedly
- Count iterations

```python
#write answer here
count=0
num = int(input("Enter the number: "))
if num==0:
    count=1
while(num!=0):
    num = num // 10
    count+=1
print("The number of digits is:",count)
```

Q7. Check if a number is palindrome.

Approach:
- Reverse number
- Compare with original

```python
#write answer here
original_num = int(input("Enter the number: "))
num = original_num
Reversed_num = 0
while(num!=0):
	Digit = num%10
	num = num//10
	Reversed_num = Reversed_num*10 + Digit
if(original_num==Reversed_num):
	print("The number is a valid palindrome")
else:
    print("Not a Valid palindrome")
```

Q8. Guess the number game.

Approach:
- Use while loop
- Continue until correct guess

```python
#write answer here
num = int(input("Enter the number: "))
guess_num = int(input("Guess the number: "))
while(guess_num!=num):
    if(guess_num<num):
        print("Too low")
    else:
        print("Too high")
    guess_num = int(input("Guess the number again: "))
print("Congratulations! You guessed it right.")
```
