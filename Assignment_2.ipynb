# Q11 - Compound assignment

- subtract 4 (`n -= 4`), print
(Place each statement so students can see the updated `n` after each operation.)

# Q16 - Input: Convert total seconds to hours, minutes, seconds

Ask the user to input a total number of seconds (integer). Convert it to hours, minutes, and seconds and print as `H:M:S`.

```python
# Answer here
# Hint: use input(), then convert to int() or float() as needed.
seconds = float(input("Enter the no. of seconds"))
print(f"Total no. of seconds : {seconds}")

minutes = seconds // 60
minutes %= 60
hours = seconds // 3600
seconds %= 60
print(f"{hours} : {minutes} : {seconds}")
```

# Q17 - Input: Simple interest calculation

Ask user for `principal`, `rate` (annual percent), and `time` (years). Compute simple interest as `(principal * rate * time) / 100` and print `Interest: <result>`.

```python
# Answer here

p = float(input("Enter principal amount : "))
print(f"Principal : {p}")
r = float(input("Enter the rate : "))
print(f"Rate : {r}%")
t = float(input("Enter the time in years : "))
print(f"Time in years : {t}")

result = (p * r * t)/100
print(f"The calculated simple Interest is {result}")
```
