### Q5
Given:
```python
A = {2, 4, 6, 8}
B = {4, 8, 12}
```
Find Union, Intersection, A-B

````python
# write your code here
A = {2,4,6,8}
B = {4,8,12}

print("The intersection of A and B is: ", A.intersection(B))
print("The union of A and B is: ", A.union(B))
print("The difference of A and B is:", A-B)
````

### Q6
Check whether 15 exists in the set {5, 10, 15, 20}.

````python
# write your code here
set_A = {5,10,15,20}
for x in set_A:
    if x == 15:
        print("The element is found")
        break
else:
    print("not found")
````
