## Question 2 (Easy)

**Title:** Sum of First N Numbers

### Problem Statement:
Take an integer N as input and find the sum of numbers from 1 to N.

### Approach:
Initialize sum = 0, loop from 1 to N, add each value to sum.


```python
# write your code here
num = int(input("Enter the number: "))
sum = 0
for i in range(1, num+1):
    sum += i
print("The sum is: ", sum)
```

## Question 6 (Easy)

**Title:** Factorial of a Number

### Problem Statement:
Take a number N as input and find its factorial.

### Approach:
Initialize fact = 1 and multiply values from 1 to N.


```python
# write your code here
num = int(input("Enter the number whose factorial you want to calculate: "))
fact = 1
if(num == 0 or num == 1):
    print("The factorial is 1")
else:
    for i in range(1,num+1):
        fact = fact * i
    print("The factorial is: ", fact)
```

## Question 9 (Easy)

Use two counters and check with modulo operator.


```python
# write your code here
count_even = 0
count_odd = 0
num = int(input("Enter the number: "))
for i in range(1, num+1):
    if i % 2 == 0:
        count_even += 1
    else:
        count_odd += 1
print("Count of even numbers: ", count_even)
print("Count of odd numbers: ", count_odd)
```

## Question 13 (Medium)

**Title:** Skip Multiples of 4

### Problem Statement:
Take N as input and print numbers from 1 to N but skip multiples of 4.

### Approach:
Use `continue` when the number is divisible by 4.


```python
# write your code here
num = int(input("Enter the number of terms: "))
for i in range(1, num+1):
    if i % 4 == 0:
        continue
    print(i, end=' ')
```

## Question 16 (Hard)

**Title:** Check Prime Number

### Problem Statement:
Take a number as input and check whether it is prime or not.

### Approach:
Loop from 2 to N-1 and use `break` if divisible.


```python
# write your code here
n = int(input("Enter the number: "))
limit = int((n**0.5)+1)
for i in range(2, limit):
    if n%i == 0:
        print(n,"is not prime")
        break
else:
    print(n,"is prime")
```

## Question 17 (Hard)

**Title:** Prime Numbers Between 1 and N

### Problem Statement:
Take N as input and print all prime numbers between 1 and N.

### Approach:
Use nested loops and a flag variable for prime checking.


```python
# write your code here
n = int(input("Enter the nth number: "))
for i in range(2,n):
    for j in range(2,int(i**0.5+1)):
        if(i%j == 0):
            break
    else:
        print(i, end=' ')

```

## Question 19 (Hard)

**Title:** Number Pattern

### Problem Statement:
Take N as input and print a right-angled number pattern using nested loops.

Example:
1
12
123

### Approach:
Use an outer loop for rows and an inner loop for printing numbers.


```python
# write your code here
num = int(input("Enter the number: "))
for i in range(1, num+1):
    for j in range(1, i+1):
        print(j, end=' ')
    print()
```
